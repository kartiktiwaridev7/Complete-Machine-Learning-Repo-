```python
import pandas as pd
from sklearn.metrics import mean_absolute_error, mean_squared_error 
```

```python
data = {
    "Coding_Hours":[1,2,3,4,5],
    "Project_Built":[1,2,3,4,5],
    "Real_Salary":[3,5,7,9,19],    #Target for Regression 
    "Got_placed":[0,0,1,1,1]  #Target for classification 
}
    
```

```python
df = pd.DataFrame(data)
```

```python
print(df.head())
```

Separate Features(X) and Target (y)

```python
x = df[["Coding_Hours" , "Project_Built"]]
```

```python
y = df["Real_Salary"]
```

Model's Predictions : Salary = 2*Coding_hours + 1

```python
predictions = [3,5,7,9,11]
```

Calculates Cost Using MAE and MSE 

```python
mae = mean_absolute_error(y,predictions)
```

```python
mse = mean_squared_error(y,predictions)
```

```python
print("Features (X) shape :" , x.shape)
```

```python
print("MAE Cost (Fair Judge): " , mae)
```

```python
print("MSE Cost (Strict Teacher):",mse)
```
